# 04 — Evaluación de resultados

Compara el modelo híbrido CNN-Transformer contra el baseline estadístico lineal (objetivo específico 4) y genera las figuras para el capítulo de resultados de la tesis.

**Requisito previo:** haber corrido `python -m src.training.train` (Fase 4) para generar `models_saved/best_model.pt`.

```python
import sys
sys.path.insert(0, '..')

import json
import matplotlib.pyplot as plt

from src.evaluation.evaluate import compare_models
from src.config import FIGURES_DIR
```

```python
# Corre la comparación completa: modelo híbrido vs. Regresión Lineal
report = compare_models()
```

## Gráfico comparativo de métricas

```python
metric_names = ["mse", "rmse", "mae"]
hybrid_values = [report["hybrid_model"][m] for m in metric_names]
baseline_values = [report["traditional_baseline"][m] for m in metric_names]

x = range(len(metric_names))
width = 0.35

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar([i - width/2 for i in x], hybrid_values, width, label="CNN-Transformer (híbrido)")
ax.bar([i + width/2 for i in x], baseline_values, width, label="Regresión Lineal (baseline)")

ax.set_xticks(list(x))
ax.set_xticklabels([m.upper() for m in metric_names])
ax.set_ylabel("Valor del error (menor es mejor)")
ax.set_title("Comparación: modelo híbrido vs. método estadístico tradicional")
ax.legend()

FIGURES_DIR.mkdir(parents=True, exist_ok=True)
fig.savefig(FIGURES_DIR / "comparacion_modelos.png", dpi=150, bbox_inches="tight")
plt.show()
```

## R² y F1-Score

```python
print("R2  -> híbrido:", round(report['hybrid_model']['r2'], 4),
      "| baseline:", round(report['traditional_baseline']['r2'], 4))
print("F1  -> híbrido:", round(report['hybrid_model']['f1'], 4),
      "| baseline:", round(report['traditional_baseline']['f1'], 4))
print("\nMejora relativa (RMSE):", round(report['relative_improvement']['rmse'] * 100, 2), "%")
print("Mejora relativa (MAE):", round(report['relative_improvement']['mae'] * 100, 2), "%")
print("Ganancia en R2:", round(report['relative_improvement']['r2_gain'], 4))
```
